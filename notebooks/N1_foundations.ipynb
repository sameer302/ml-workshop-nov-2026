## 1. How do computers work?
A computer is a machine that **takes input, processes it, and gives output**.

![Input, process, output](https://raw.githubusercontent.com/sameer302/ml-workshop-nov-2026/main/images/n1_input_process_output.png)

*Example: you type a message (input), the phone works on it (process), the text appears on the screen (output).*

## 2. How do we give input to a computer?
Inside a computer, everything is built from switches that are either **OFF (0)** or **ON (1)**.

- **Bit** = one 0 or 1
- **Byte** = 8 bits
- Letters, numbers, photos, songs: all stored as patterns of 0s and 1s

**Characters become numbers first.** Every character is given a number, and that number is stored in bits.
- **ASCII**: the original table, 128 characters (English letters, digits, symbols). `A` = 65, `a` = 97
- **Unicode**: the modern table, with well over 150,000 characters from all languages, symbols and emoji. Its first 128 characters are the same as ASCII

```python
for ch in ["A", "a", "7", "अ", "😀"]:
    number = ord(ch)                       # the character's number (Unicode code point)
    print(ch, "-> number", number, "-> bits", format(number, "b"))

print("The number 25 in binary is", format(25, "b"))

print()
print("So to talk to a computer directly, we would have to talk in 0s and 1s.")
print("Imagine if humans had to talk like that:")
print()
print("[ Instructor plays the clip here ]")
```

## 3. So we invented programming languages
Nobody wants to write 0s and 1s. So we built languages that are easier for humans.

![Language ladder](https://raw.githubusercontent.com/sameer302/ml-workshop-nov-2026/main/images/n1_language_ladder.png)

But the CPU only understands machine code. So any higher-level code is eventually converted to machine code by a **translator program**.

*Example: Python first translates our code into a lower-level form called bytecode, and then runs it.*

```python
import dis

def add(a, b):
    return a + b

dis.dis(add)      # the lower-level instructions Python made from our code
print(add(2, 3))  # and then it runs them
```

## 4. How we have always made applications
Until now, we write a program in a language like Python. The program is a set of **rules** written by a human.

![Traditional programming](https://raw.githubusercontent.com/sameer302/ml-workshop-nov-2026/main/images/n1_traditional_programming.png)

This is how we made calculators, websites, mobile apps and games: the human works out the rules, and the computer follows them exactly.

## 5. An easy task: pass or fail?
Task: given a student's marks, decide **Pass** or **Fail**. Say the pass mark is 40.

We know the rule, so writing the program is easy.

```python
def result(marks):
    if marks >= 40:
        return "Pass"
    else:
        return "Fail"

for m in [85, 40, 39, 12]:
    print(m, "->", result(m))
```

## 6. A harder task: cat or dog?
Task: given a photo, decide **cat** or **dog**. It is also a two-way decision, just like pass or fail.

| ![Cat](https://raw.githubusercontent.com/sameer302/ml-workshop-nov-2026/main/images/n1_cat.jpg) | ![Dog](https://raw.githubusercontent.com/sameer302/ml-workshop-nov-2026/main/images/n1_dog.jpg) |
|:---:|:---:|
| **Cat** | **Dog** |

You can tell them apart in an instant. But can we write the rules?

| A rule we might try | Why it breaks |
|---|---|
| Cats have pointy ears | Many dogs have pointy ears, and some cats have folded ears |
| Dogs are bigger | A large cat can be bigger than a tiny dog |
| Cats have whiskers | Dogs have whiskers too |
| Check the fur colour | Both come in many colours and patterns |

On top of that, every photo is different: angle, light, background, and parts of the animal hidden.

**And the computer does not see "ears" or "whiskers". It sees only a grid of numbers.**

## 7. Easy for us, hard to program
We recognise these in an instant, but we cannot explain the exact steps we follow:
- Recognising a friend's face
- Understanding speech in a noisy room
- Reading someone's handwriting
- Spotting a problem in a medical image

We know the answer when we see it, but we cannot write the rules down. So the "write the rules by hand" approach hits a wall.

## 8. Flip the idea
What if we do not write the rules? What if we **show the computer many examples** and let it **find the rules itself**?

![Traditional programming vs machine learning](https://raw.githubusercontent.com/sameer302/ml-workshop-nov-2026/main/images/n1_traditional_vs_ml.png)

For cat or dog: show thousands of photos, each labelled "cat" or "dog", and let the computer work out what separates them.

## 9. Enter AI and machine learning
- **Artificial Intelligence (AI):** computer systems that do tasks which normally need human intelligence, like recognising images, understanding language, reasoning and making decisions
  - Not every program is AI: adding numbers or checking pass or fail is just calculation
  - AI can be built with hand-written rules, or by letting the computer learn
- **Machine Learning (ML):** the part of AI where the computer **learns from data** instead of being given rules
- **Deep Learning (DL):** a kind of ML that uses artificial neural networks with many layers, loosely inspired by the human brain

![AI, ML and DL](https://raw.githubusercontent.com/sameer302/ml-workshop-nov-2026/main/images/n1_ai_ml_dl.png)

## 10. Recap, and what comes next
- Traditionally, **a human writes the rules** and the computer follows them
- That works when we can state the rules (pass or fail)
- It breaks when we cannot (cat or dog)
- Machine learning flips the idea: **the computer learns the rules from examples**
