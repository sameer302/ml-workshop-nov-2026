## 1. What is a computer?
A computer is a machine that **takes input, processes it, and gives output**.

![Input, process, output](https://raw.githubusercontent.com/sameer302/ml-workshop-nov-2026/main/images/n1_input_process_output.png)

*Example: you type a message (input), the phone works on it (process), the text appears on the screen (output).*

## 2. How do we give input to a computer?
A computer cannot understand English, Hindi or pictures directly.
Inside, everything is built from switches that are either **OFF (0)** or **ON (1)**.

- **Bit** = one 0 or 1
- **Byte** = 8 bits
- Letters, numbers, photos, songs: all stored as patterns of 0s and 1s

So to talk to a computer directly, we would have to talk in **0s and 1s**.

## 3. Bits and bytes in action
Every character has a number, and every number can be written in bits.

```python
text = "Hi"
for ch in text:
    print(ch, "-> number", ord(ch), "-> bits", format(ord(ch), "08b"))

print("The number 25 in binary is", format(25, "b"))
```

## 5. So we invented programming languages
Nobody wants to write 0s and 1s. So we built languages that are easier for humans, and tools that translate them for the computer.

![Language ladder](https://raw.githubusercontent.com/sameer302/ml-workshop-nov-2026/main/images/n1_language_ladder.png)

## 6. Low-level vs high-level languages
| | Low-level | High-level |
|---|---|---|
| Examples | Machine code, Assembly | C, C++, Python |
| Close to | The hardware | Human thinking |
| Ease of writing | Hard | Easier |
| Control over hardware | Very high | Less |
| Same task, adding 2 + 3 | `mov eax, 2` then `add eax, 3` (assembly) | `a = 2 + 3` |

Higher-level languages let us write less and think more about the problem.

## 7. From source code to a running program
What we write is called **source code**. The CPU only runs **machine code**. A translator sits in between.

![Compile and interpret paths](https://raw.githubusercontent.com/sameer302/ml-workshop-nov-2026/main/images/n1_translation_pipeline.png)

## 8. Compiler vs interpreter
| | Compiler | Interpreter |
|---|---|---|
| Translates | The whole program, once, before running | One statement at a time, while running |
| Output | Machine code file (object code, then executable) | No separate file |
| Speed when running | Usually faster | Usually slower |
| Finding errors | Many errors found before the program runs | Errors appear when that line is reached |
| Examples | C, C++ | Python |

**Analogy:** a compiler is like translating a whole book before anyone reads it. An interpreter is like a live translator at a speech.

## 9. Python in action
In practice, many languages mix both ideas. The standard Python (CPython) first compiles your code to **bytecode**, then its interpreter runs that bytecode.

```python
import dis

def add(a, b):
    return a + b

dis.dis(add)      # the bytecode Python made from our source code
print(add(2, 3))  # the interpreter runs it
```
