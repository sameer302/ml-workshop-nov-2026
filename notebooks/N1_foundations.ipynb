## 1. What is a computer?
A computer is a machine that **takes input, processes it, and gives output**.

![Input, process, output](https://raw.githubusercontent.com/sameer302/ml-workshop-nov-2026/main/images/n1_input_process_output.png)

*Example: you type a message (input), the phone works on it (process), the text appears on the screen (output).*

## 2. How do we give input to a computer?
Inside a computer, everything is built from switches that are either **OFF (0)** or **ON (1)**.

- **Bit** = one 0 or 1
- **Byte** = 8 bits
- Letters, numbers, photos, songs: all stored as patterns of 0s and 1s

**Characters become numbers first.** Every character is given a number, and that number is stored in bits.
- **ASCII**: the original table, 128 characters (English letters, digits, symbols). `A` = 65, `a` = 97
- **Unicode**: the modern table, with well over 150,000 characters from all languages, symbols and emoji. Its first 128 characters are the same as ASCII

```python
from IPython.display import YouTubeVideo

for ch in ["A", "a", "7", "अ", "😀"]:
    number = ord(ch)                       # the character's number (Unicode code point)
    print(ch, "-> number", number, "-> bits", format(number, "b"))

print("The number 25 in binary is", format(25, "b"))

print()
print("So to talk to a computer directly, we would have to talk in 0s and 1s.")
print("Imagine if humans had to talk like that:")

YouTubeVideo("ngDF99ETb2Y", width=720, height=405)
```

## 3. So we invented programming languages
Nobody wants to write 0s and 1s. So we built languages that are easier for humans.

![Language ladder](https://raw.githubusercontent.com/sameer302/ml-workshop-nov-2026/main/images/n1_language_ladder.png)

But the CPU only understands machine code. So any higher-level code is eventually converted to machine code by a **translator program**.

*Example: Python first translates our code into a lower-level form called bytecode, and then runs it.*

```python
import dis

def add(a, b):
    return a + b

dis.dis(add)      # the lower-level instructions Python made from our code
print(add(2, 3))  # and then it runs them
```
